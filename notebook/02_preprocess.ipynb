```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.model_selection import train_test_split

print("Imports successful ✓")
```

```python
dataset_path = "../datasets"

data = []
for filename in os.listdir(dataset_path):
    if filename.endswith(".jpg"):
        parts = filename.split("_")
        if len(parts) >= 3:
            try:
                age = int(parts[0])
                gender = int(parts[1])
                if gender in [0, 1]:  # remove gender=3 bad entry
                    data.append({
                        "filename": filename,
                        "age": age,
                        "gender": gender
                    })
            except:
                pass

df = pd.DataFrame(data)

def get_age_group(age):
    if age <= 12:   return 0  # Child
    elif age <= 19: return 1  # Teen
    elif age <= 35: return 2  # Young Adult
    elif age <= 55: return 3  # Adult
    else:           return 4  # Senior

df["age_group"] = df["age"].apply(get_age_group)

print(f"Clean dataset size: {len(df)}")
print(f"Age groups:\n{df['age_group'].value_counts().sort_index()}")
```

```python
IMG_SIZE = 64  # resize all images to 64x64

images = []
genders = []
age_groups = []

for _, row in df.iterrows():
    img_path = os.path.join(dataset_path, row["filename"])
    try:
        img = Image.open(img_path).convert("RGB")
        img = img.resize((IMG_SIZE, IMG_SIZE))
        images.append(np.array(img))
        genders.append(row["gender"])
        age_groups.append(row["age_group"])
    except:
        pass

# Convert to numpy arrays
X = np.array(images, dtype="float32") / 255.0  # normalize 0-1
y_gender = np.array(genders)
y_age = np.array(age_groups)

print(f"X shape: {X.shape}")
print(f"y_gender shape: {y_gender.shape}")
print(f"y_age shape: {y_age.shape}")
print("Preprocessing complete ✓")
```

```python
X_train, X_test, yg_train, yg_test, ya_train, ya_test = train_test_split(
    X, y_gender, y_age,
    test_size=0.2,
    random_state=42
)

print(f"Training samples : {len(X_train)}")
print(f"Testing samples  : {len(X_test)}")
```

```python
np.save("../models/X_train.npy", X_train)
np.save("../models/X_test.npy", X_test)
np.save("../models/yg_train.npy", yg_train)
np.save("../models/yg_test.npy", yg_test)
np.save("../models/ya_train.npy", ya_train)
np.save("../models/ya_test.npy", ya_test)

print("All data saved to models/ folder ✓")
```
